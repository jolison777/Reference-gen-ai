```python
def start_message(func):   #defined a function(decorator)
    def wrapper():
        print("Function begin") #printed a statement
        func()
    return wrapper

@start_message    #function started
def login():
    print("Login successful")

@start_message      #functionstate
def logout():
    print("Logout successful")
#do start message here
login()
logout()
```

```python
# *args
def ar(*args):
    print(args)   #WE CAN SPECIFICALLY PRINT USING INDEX

ar(10,20,30,40,50,60)   #tuple
```

```python

def details(**kwargs):
    print(kwargs) #kwrgs can print specific value

details(name="TAMIZH", age=22, city="Chennai",pincode=600100)
```

```python
def display(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

display(10, 20, 30, name="MJ", city="Chennai")
```

```python
def countdown(n): #recurssion with one argument
  if n <= 1:    #checking for  condition
    print("done")
  else:
    print(n)      #5
    countdown(n-1)   #calling the same function inside itself

countdown(6)   #function called
```

```python
def demo():
    print("Start")
    yield 1    #checkpoint

    print("Middle")
    yield 2

    print("End")
    yield 3

g = demo()

print(next(g))#The next time next() is called, it continues from where it stopped.
print(next(g))

```
